### Handling Missing values of the age column

```python
# Finding the difference between mean and median age

print (df["age"].mean())
print (df["age"].median())
print (df["age"].mean() == df["age"].median()) # True if both are equal
```

```python
# Filling missing values of age with the median
df["age"] = df["age"].fillna(df["age"].median()).astype(int)

# Verify
print (df["age"].isnull().sum())
```

### Handling Missing values in the Years_of_experience column

```python
# Finding the mean and median for the years_of_experience column
print (df["years_of_experience"].mean())
print (df["years_of_experience"].median())
```

```python
# Replacing missing values with the median value
df["years_of_experience"] = df["years_of_experience"].fillna(df["years_of_experience"].median()).astype(int)

# Verify
print (df["years_of_experience"].isnull().sum())
```

### Handling Missing values of the hourly_rate (USD) column


```python
# Check unique values and data type of the hourly_rate (USD) column
print (df["hourly_rate (USD)"].unique())
print (df["hourly_rate (USD)"].dtype)
```

### Standardize Hourly Rate

The "hourly_rate (USD)" column contains inconsistent string formats with currency symbols(such as "USD" and "$"), plus missing values. To standardize this column for analysis :

1. Remove the currency text markers.

2. Convert the column into a uniform numeric data type.

3. Fill any missing values using an appropriate statistical imputation method.





```python
# Remove "USD" and "$" text
df["hourly_rate (USD)"] = df["hourly_rate (USD)"].astype(str).str.replace("USD","").str.replace("$","").str.strip()
df["hourly_rate (USD)"] = pd.to_numeric(df["hourly_rate (USD)"], errors="coerce")
```

```python
# Finding mode for this column
print (df["hourly_rate (USD)"].mode())
```

```python
# Replacing missing values in the hourly_rate (USD) column with the mode
df["hourly_rate (USD)"] = df["hourly_rate (USD)"].fillna(40)

# Verify
print (df["hourly_rate (USD)"].isnull().sum())
print (df["hourly_rate (USD)"].dtype)
```

### Handling Missing values of the rating column

```python
# Finding the mean and median
print (df["rating"].mean())
print (df["rating"].median())
```

```python
# Replacing the  missing values with the median
df["rating"] = df["rating"].fillna(df["rating"].median())

# Verify
print (df["rating"].isnull().sum())
print (df["rating"].dtype)
```

### Handling Missing values of the is_active column

```python
# Checking unique values
print(df["is_active"].unique())
```

### Standardize Inconsistent Categories

The unique value check reveals mixed data representations (strings,booleans,and numbers etc..). Standardizing these variations into a uniform numeric format ensures accurate mode calculation and prevents errors during analysis.

```python
# Standardize the Categories
df["is_active"] = df["is_active"].astype(str).str.strip().str.lower()

df["is_active"] = df["is_active"].replace({
    "no":0,"false":0,"0":0,"0.0":0,"n":0,"f":0,
    "yes":1,"true":1,"1":1,"1.0":1,"y":1,"t":1
})

```

```python
# Impute missing values
df["is_active"] = pd.to_numeric(df["is_active"], errors="coerce").fillna(0).astype(int)
df["is_active"] = df["is_active"].map({1:"Yes",0:"No"})

# Verify
print(df["is_active"].unique())
```

### Handling Missing values of the client_satisfaction column

```python
# Check unique values and data type of client_satisfaction
print(df["client_satisfaction"].unique())
print(df["client_satisfaction"].dtype)
```

### Standardize and Handle Missing Values for client_satisfaction

The "client_satisfaction" column contains inconsistent string representations(such as percentage symbols like "84%",plain numbers, and "nan" values. To prepare this data for analysis:

1. Remove the percentage signs and convert the column into a uniform numeric format.
2. Coerce any invalid entries to missing values("nan").

  

3. Impute the missing values using the median value of the column.

```python
# Remove the "%" sign and convert to numeric
df["client_satisfaction"] = df["client_satisfaction"].astype(str).str.rstrip("%")

# Coerce any invalid entries to missing values
df["client_satisfaction"] = pd.to_numeric(df["client_satisfaction"], errors="coerce")

# Impute missing values with the median
df["client_satisfaction"] = df["client_satisfaction"].fillna(df["client_satisfaction"].median()).astype(int)
print(df["client_satisfaction"].median())

# Verify
print(df["client_satisfaction"].unique())
print(df["client_satisfaction"].dtype)
```

### Final Check for Missing Values

```python
df.info()
```

All missing values have been filled.

### Standardize Gender Column

The "gender" column contains inconsistent string variations (such as "FEMALE", "female","F",and "male"). To standardize this column for consistent analysis :

1. Inspect all unique values and frequencies to identify variations.

2. Normalize the variations into uniform categories.

```python
# Check unique values and datatype of the gender column

print (df["gender"].unique())
print (df["gender"].value_counts())
```

```python
# Standardizing
df["gender"] = df["gender"].astype(str).str.strip().str.lower()

df["gender"] = df["gender"].replace({
    "f": "Female",
    "female":"Female",
    "m": "Male",
    "male":"Male",
})

# Verify
print (df["gender"].unique())
print(df["gender"].value_counts())
```

### Key Data Cleaning & Preprocessing Steps





*   Age & Experience :  Calculated descriptive statistics(mean and median) and filled missing entries using the median value.
*   Hourly Rate          :  Removed inconsistent string markers (such as "USD" and "$"), Converted the column into a uniform numeric data type,and handled missing values using the mode.


*   Rating               : Analyzed the distribution using mean and median metrics before filling missing values with the median.
*   Is_Active            : Inspected unique categories, standardized inconsistent text representations and imputed missing data.

*   Client Satisfaction  : Cleaned mixed string formats containing percentages,plain numbers,and missing entries, converted them to numeric formats, and imputed missing values using the median.
*   Gender               : Inspected unique string variations(such as "FEMALE","F","female",and "male") and normalized them into uniform categories("Female" and "Male").

### **Visualizations**

This section utilizes a variety of plots to uncover key trends,distributions,and relationships within the dataset.

```python
# Line Chart
# Group data by experience to show a trend
grouped_data = df.groupby("years_of_experience")["hourly_rate (USD)"].mean().reset_index()
plt.figure(figsize=(8,5))
sns.set_style("whitegrid")
sns.lineplot(x="years_of_experience",y="hourly_rate (USD)",data=grouped_data,color="#1f77b4")
plt.title("Average Hourly Rate by Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Average Hourly Rate (USD)")
plt.show()
```

Analysis of Average Hourly Rate by years of experience

The line chart illustrates the relationship between a freelancers years of experience and their average hourly rate.


*   Early Career (0-15 Years) : Hourly rates stay steady, mostly ranging between 45 and 55.
*   Mid-to-Late Career(16-30 Years) : Rates fluctuate more.Notably, the highest average rate peak appears near the mid-twenties experience mark,reaching approximately 71/hr.


*   Advanced Experience(35+ Years) : The sharp drops and spikes happen because there are fewer freelancers with that much experience,making the averages more unstable.
